# AgriX AI: Exploratory Data Analysis (EDA) for Crop Disease Dataset

This notebook inspects the raw image dataset, visualizes sample leaf lesions per class, checks image dimensions, and evaluates class balance.

```python
import os
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image
import pandas as pd
import yaml

with open('../configs/training_config.yaml', 'r') as f:
    config = yaml.safe_load(f)

raw_dir = Path('../data/raw')
classes = config['dataset']['classes']
print(f"Target Scope: {len(classes)} Classes")
```

```python
# Class distribution audit
class_counts = {}
for c in classes:
    class_folder = raw_dir / c
    if class_folder.exists():
        count = len([f for f in class_folder.iterdir() if f.is_file()])
        class_counts[c] = count
    else:
        class_counts[c] = 0

df_counts = pd.DataFrame(list(class_counts.items()), columns=['Class', 'Count'])
plt.figure(figsize=(12, 6))
plt.barh(df_counts['Class'], df_counts['Count'], color='teal')
plt.xlabel('Number of Images')
plt.title('AgriX Class Distribution')
plt.tight_layout()
plt.show()
```
